```python
print(f'(a):\n')

x1 = [5, 15, 15, 25]  # average speed deviation (km/h)
x2 = [10, 30, 50, 30] # number of stops
x3 = [5, 15, 15, 5]   # idle time (minutes)

n = 4

mean_x1 = sum(x1) / n
mean_x2 = sum(x2) / n
mean_x3 = sum(x3) / n

variance_x1 = []
for num in x1:
    variance_x1.append((num - mean_x1)**2)
    
variance_x2 = []
for num in x2:
    variance_x2.append((num - mean_x2)**2)
                       
variance_x3 = []
for num in x3:
    variance_x3.append((num - mean_x3)**2)

s_squared_x1 = sum(variance_x1) / n
s_squared_x2 = sum(variance_x2) / n  
s_squared_x3 = sum(variance_x3) / n 

print(f'x1 = {x1} -> mean = {mean_x1} | squared deviations = {variance_x1} | variance = {s_squared_x1}')
print(f'x2 = {x2} -> mean = {mean_x2} | squared deviations = {variance_x2} | variance = {s_squared_x2}')
print(f'x3 = {x3} -> mean = {mean_x3} | squared deviations = {variance_x3} | variance = {s_squared_x3}')
             
print(f'\n(b):\n')

total_variance = s_squared_x1 + s_squared_x2 + s_squared_x3
print(f'the total variance of the dataset = {total_variance}')
```

```python
print(f'(c):\n')

#       w1   w2  w3
PC1 = [0.6, 0.8, 0]

#          x1   x2   x3
driver_1 = [5,  10,  5]
driver_2 = [15, 30,  15]
driver_3 = [15, 50,  15]
driver_4 = [25, 30,  5]

i = 0
projection_1 = 0
for num in driver_1:
    projection_1 += num * PC1[i]
    i += 1

i = 0
projection_2 = 0
for num in driver_2:
    projection_2 += num * PC1[i]
    i += 1


projection_3 = 0
for num, weight in zip(driver_3, PC1):
    projection_3 += num * weight


projection_4 = 0
for num, weight in zip(driver_4, PC1):
    projection_4 += num * weight

# חישבתי את ההיטל בשתי דרכים בקוד: התאמה לפי מיקום ברשימה והתאמה ישירה בין איברי הרשימות. אלו שתי דרכי מימוש של אותו חישוב מתמטי

print(f'driver_1 = {driver_1} | projection_1 = {projection_1}')
print(f'driver_2 = {driver_2} | projection_2 = {projection_2}')
print(f'driver_3 = {driver_3} | projection_3 = {projection_3}')
print(f'driver_4 = {driver_4} | projection_4 = {projection_4}')

total = projection_1 + projection_2 + projection_3 + projection_4
n = 4
mean = total / n

print(f'\ntotal = {total} | n = {n} | average = {mean}')

deviations = []
for num in [projection_1, projection_2, projection_3, projection_4]:
    deviations.append(num - mean)

deviations_squared = []
for num in deviations:
    deviations_squared .append(num**2)

s_squared_pc1 = sum(deviations_squared) / n   

ratio = s_squared_pc1 / total_variance

print(f'\ndeviations = {deviations} | deviations_squared = {deviations_squared}')

print(f'\ns_squared_pc1 = {s_squared_pc1} | total_variance = {total_variance} | ratio = {ratio:.4f}\n')

print('העמודה החדשה שנוצרת משילוב שלוש העמודות המקוריות, מסבירה 70.55% מהשונות בנתונים')
```
