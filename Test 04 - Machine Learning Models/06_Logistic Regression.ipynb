```python
print(f'-----Qustion 8-----:\n')

import math
e = math.e 

x = 3
z = -3 + (0.6 * x)
p = 1 / (1 + e**(-z))
print(f'e = {e:.3f}, z = {z:.3f}, p = {p:.3f}')
print(f'the probability of spam is approximately {p*100:.2f}%.')
```

```python
print(f'----Bonus Qustion-----:\n')

import math

p = 0.8
log_odds = math.log(p / (1 - p))

x_80 = (log_odds + 3) / 0.6
print(f'at 80% probability, the log-odds are {log_odds:.3f}')
print(f'\nSolving -3 + 0.6x = {log_odds:.3f} gives x ≈ {x_80}')

# Check the result

e = math.e 
z = -3 + 0.6 * x_80
checked_p = 1 / (1 + e**(-z))


print(f'\nChecking: z = {z:.3f}, probability = {checked_p:.2%}.')
print(f'\nThere is no whole number of keywords that gives exactly 80%.')
```
