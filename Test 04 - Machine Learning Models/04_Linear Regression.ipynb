```python
print(f'-----Qustion 4-----\n(a):\n')
#y_hat = 6*x +10

y_hat_9 = 6 * 9 + 10
print(f'the predict score of a student who studied for 9 hours is -> {y_hat_9}')

print(f'\n(b):\n')

y_hat_5 = 6 * 5 +10
e = 41 - y_hat_5
print(f'the residual for the student with x = 5 is -> {e}')


print(f'\n(c):\n')

x_hours_studied = [1, 3, 5, 7]
y_quiz_score = [19, 25, 41, 51]


for hours, actual_score in zip(x_hours_studied, y_quiz_score):
    y_hat = 6 * hours + 10
    e = actual_score - y_hat

    print(f'x={hours}, y={actual_score}, y_hat={y_hat}, e={e}, e²={e**2}')
    
import numpy as np

e = [3, -3, 1, -1]
sse_list = []
for num in e:
    sse_list.append(num**2)

sse = np.sum(sse_list)
print(f'\nSSE = {sse}')
MSE = sse / len(x_hours_studied)
print(f'MSE = {MSE}')

RMSE = MSE ** 0.5
print(f'RMSE = {RMSE}')


print(f'\n(d):\n')  

mean = np.sum(y_quiz_score) / len(y_quiz_score)

sst_list = []
for num in y_quiz_score:
    sst_list.append((num-mean)**2)

sst = np.sum(sst_list)
print(f'SST = {sst}')

r_2 = 1 - (sse / sst)
print(f'R2 = {r_2}')
print(f'The model explains {r_2:.1%} of the variation in quiz scores, the remaining {1 - r_2:.1%} is everything temperature does not capture.')
```
