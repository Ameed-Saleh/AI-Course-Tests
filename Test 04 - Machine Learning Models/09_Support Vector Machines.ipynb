```python
print(f'----Quetion 13----\n')
print(f'(a) + (b):\n')
# Boundary: 6*x1 + 8*x2 - 48 = 0
# P = (10, 2), Q = (3, 3)

P_value = 6 * 10 + 8 * 2 - 48
Q_value = 6 * 3 + 8 * 3 - 48
norm_w = (6**2 + 8**2) ** 0.5

distance_P = P_value / norm_w
distance_Q = Q_value / norm_w

print(f'P: value = {P_value} => positive side -> distance = {distance_P:.1f}')
print(f'\nQ: value = {Q_value} => negative side -> distance = {distance_Q:.1f}')

print(f'\n(c)')
print(f'Q is closer to the boundary: {distance_Q:.1f} units versus {distance_P:.1f} units for P.')
```
