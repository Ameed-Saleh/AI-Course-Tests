```python
min_x = 18
max_x = 90
denominator = max_x - min_x

for num in [18, 42, 66, 90]:
    print(f'x_norm_{num} = {(num - min_x) / denominator}')
```

```python
import numpy as np

numbers = [8, 10, 10, 14, 14, 16]

mean = np.sum(numbers) / len(numbers)
print(f"The mean is: {mean}")

deviations_from_mean = []
for number in numbers:
    deviations_from_mean.append(number - mean)

squared_deviations = []
for deviation in deviations_from_mean:
    squared_deviations.append(deviation ** 2)

print(f'Deviations from mean: {deviations_from_mean}')
print(f'Squared deviations: {squared_deviations}')

sum_squared_deviations = np.sum(squared_deviations)
variance = sum_squared_deviations / len(numbers)
standard_deviation = variance ** 0.5

print(f'Standard deviation: {standard_deviation}')
```
